```python
import matplotlib.pyplot as plt
```

```python
# Plot the bar chart
plt.bar(df['Name'], df['Salary'])

# Add labels and title
plt.xlabel('Name')
plt.ylabel('Salary ($)')
plt.title('Salary by Employee')

# Show the chart
plt.show()
```

```python
# Plot the bar chart (red color)
plt.bar(df['Name'], df['Salary'], color='red')

# Add labels and title
plt.xlabel('Name')
plt.ylabel('Salary ($)')
plt.title('Salary by Employee (Red Bars)',fontsize=30)

# Display the chart
plt.show()
```

```python
# Create a table figure
fig, ax = plt.subplots()
ax.axis('off')  # Hide axes

# Create the table
table = ax.table(
    cellText=df.values,
    colLabels=df.columns,
    loc='center',
    cellLoc='center'
)

# Adjust layout
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1.2, 1.2)

# Add title
plt.title("Employee Data Table", fontsize=14, fontweight='bold')

# Show the table
plt.show()
```

```python

# Create the table figure
fig, ax = plt.subplots()
ax.axis('off')  # Hide axes

# Create the table
table = ax.table(
    cellText=df.values,
    colLabels=df.columns,
    loc='center',
    cellLoc='center'
)

# Adjust layout and font
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1.2, 1.2)

# Color the header row red with white text
for (row, col), cell in table.get_celld().items():
    if row == 0:  # Header row
        cell.set_facecolor('red')
        cell.set_text_props(color='white', fontweight='bold')

# Add title
plt.title("Employee Data Table (Red Headers)", fontsize=14, fontweight='bold')

# Show the table
plt.show()
```
